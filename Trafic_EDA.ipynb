```python
import matplotlib.pyplot as plt
import pandas as pd
import os

# Safely verify if df exists, otherwise locate and reload it
global_vars = globals()
if 'df' not in global_vars:
    csv_file = [f for f in os.listdir('.') if f.endswith('.xls') or f.endswith('.csv')]
    if csv_file:
        df = pd.read_csv(csv_file[0])
    else:
        # Fallback to loading the standard dataset if defined in parent steps
        try:
            df = pd.read_csv('29162b87-f1a7-4011-b309-b14d904f9a7d.csv.xls')
        except Exception:
            df = pd.read_csv('/content/29162b87-f1a7-4011-b309-b14d904f9a7d.csv.xls')

categories = [
    "Fatal Accidents",
    "Grievous Injury Accidents",
    "Minor Injury Accidents",
    "Non-Injury Accidents"
]

totals = [
    df["Fatal Accidents"].sum(),
    df["Grievous Injury - Accidents"].sum(),
    df["Minor Injury - Accidents"].sum(),
    df["Non Injury"].sum()
]

plt.figure(figsize=(10, 6))

plt.bar(categories, totals)

plt.title("Accident Category Comparison (1998–2025)")
plt.xlabel("Accident Category")
plt.ylabel("Total Count Across Years")

plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

category_summary = pd.DataFrame({
    "Accident Category": categories,
    "Total Count": totals
})

category_summary = category_summary.sort_values(
    by="Total Count",
    ascending=False
)

print(category_summary)
```

```python
# Calculate year-over-year change
df["Accident Change"] = df["Total Accidents"].diff()

df["Accident Change %"] = (
    df["Total Accidents"].pct_change() * 100
)

# Display results
print(df[[
    "YEAR",
    "Total Accidents",
    "Accident Change",
    "Accident Change %"
]].round(2))

# Highest percentage increase
increase = df.loc[df["Accident Change %"].idxmax()]

print("\nHighest Percentage Increase:")
print("Year:", int(increase["YEAR"]))
print("Change:", round(increase["Accident Change %"], 2), "%")

# Highest percentage decrease
decrease = df.loc[df["Accident Change %"].idxmin()]

print("\nHighest Percentage Decrease:")
print("Year:", int(decrease["YEAR"]))
print("Change:", round(decrease["Accident Change %"], 2), "%")
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.bar(df["YEAR"], df["Accident Change %"])

plt.axhline(y=0, linewidth=1)
plt.title("YEAR-over-Year Change in Total Accidents")
plt.xlabel("YEAR")
plt.ylabel("Change (%)")

plt.xticks(df["YEAR"], rotation=90)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Select correct numeric columns for correlation based on actual DataFrame columns
corr_cols = [
    "Fatal Accidents",
    "Fatal Killed",
    "Grievous Injury - Accidents",
    "Grievous Injurty - Injured",
    "Minor Injury - Accidents",
    "Minor Injury - Injured",
    "Non Injury",
    "Total Accidents",
    "Total number of persons involved"
]

corr = df[corr_cols].corr()

# Display correlation values
print(corr.round(2))
```

```python

plt.figure(figsize=(11, 8))

plt.imshow(corr, vmin=-1, vmax=1, aspect="auto")

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(corr_cols)),
    corr_cols,
    rotation=90
)

plt.yticks(range(len(corr_cols)), corr_cols)

plt.title("Traffic Accident Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Find highest and lowest accident years

highest_year = df.loc[df["Total Accidents"].idxmax()]
lowest_year = df.loc[df["Total Accidents"].idxmin()]

print("TRAFFIC ACCIDENT ANALYSIS - KEY FINDINGS")
print("-" * 45)

print(
    f"Highest accidents: {highest_year['Total Accidents']:,} "
    f"in {int(highest_year['YEAR'])}"
)

print(
    f"Lowest accidents: {lowest_year['Total Accidents']:,} "
    f"in {int(lowest_year['YEAR'])}"
)

print(
    f"Average accidents per year: "
    f"{df['Total Accidents'].mean():,.2f}"
)

print(
    f"Average fatalities per year: "
    f"{df['Fatal Killed'].mean():,.2f}"
)

print(
    f"Average persons involved per year: "
    f"{df['Total number of persons involved'].mean():,.2f}"
)
```
